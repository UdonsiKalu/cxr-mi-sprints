## 6. Run


```python
from _common import top_logits, encode, device, token_ids
import torch

ids = encode(tok, PROMPT_TEST, device(model))
with torch.inference_mode():
    out = model(**ids)
logits = out.logits[0, -1, :].float().cpu()
yes_ids = token_ids(tok, (" Yes", "Yes", " yes"))
no_ids = token_ids(tok, (" No", "No", " no"))
yes_max = max(float(logits[i]) for i in yes_ids)
no_max = max(float(logits[i]) for i in no_ids)
print("PROMPT_TEST:", PROMPT_TEST)
print("yes_max", yes_max, "no_max", no_max, "margin yes-no", yes_max - no_max)
look.cmd_logits(model, tok, PROMPT_TEST)
```
